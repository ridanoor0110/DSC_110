What do we do with missing data values?
1. dropping rows
2. fill in with the mean

```python
#drop all rows that have any missing values
df_complete = df.dropna()
df_complete.shape[0]
```

```python
#drop only the rows where select fields have missing values
df_complete_bmi = df.dropna(subset = ['bmi'])
df_complete_bmi.shape[0]
```

```python
#filling na with mean, zero or any other variable
#get mean and assign it to a variable
mean_bp = df['systolic_bp'].mean()
print (mean_bp)

df['systolic_bp'] = df['systolic_bp'].fillna(mean_bp)
```

```python
#verify that the na fields have been replaced with the mean for systolic_Bp
df['systolic_bp'].isna().sum()
```

transforming variables

```python
def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "underweight"
  elif bmi < 25:
    bmi_result = "normal"
  elif bmi < 30:
    bmi_result = "overweight"
  else:
    bmi_result = "obese"
  return bmi_result

df['bmi_category'] = df['bmi'].apply(bmi_check)
#applies the bmi check function to our dataset for the bmi field

#verify it with value_counts()

df['bmi_category'].value_counts()
#counts the number of output for each category indicated by the function
```

```python
#new function called "cut"
#does same thing as earlier but summarizes in much less lines of code

df['bmi_category2'] = pd.cut(df['bmi'],
                             bins = [0, 18.5, 25, 30, 100],
                             labels = ['underweight', 'normal', 'obese', 'overweight'])

#verify it
df['bmi_category2'].value_counts()
```

```python
#see missing values
df['bmi_category2'].value_counts(dropna = False)
```
